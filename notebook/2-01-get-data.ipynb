## 2-1. データ収集

```python
import pandas as pd
import os
import urllib.request
```

```python
dfs = []

for index in range(16):
    url = f"https://raw.githubusercontent.com/waico/SKAB/master/data/valve1/{index}.csv" 
    file_path = f'../data/unprocessed-data/valve1_{index}.csv'
    if not os.path.exists(file_path):
        urllib.request.urlretrieve(url, file_path)
    df = pd.read_csv(file_path, sep=';')
    dfs.append(df)

# データの結合
df = pd.concat(dfs, ignore_index=True)
print(f"\n結合後のデータ: {len(df)} rows")
df.head(3)

df.to_csv('../data/unprocessed-data/valve1.csv', index=False)
```
