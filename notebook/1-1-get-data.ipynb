## 1-1. データ収集

```python
import pandas as pd
import os
import urllib.request
```

```python
data_urls = {
    'valve1_1': 'https://raw.githubusercontent.com/waico/SKAB/master/data/valve1/1.csv',
    'valve1_2': 'https://raw.githubusercontent.com/waico/SKAB/master/data/valve1/2.csv',
    'valve1_3': 'https://raw.githubusercontent.com/waico/SKAB/master/data/valve1/3.csv',
    'valve1_4': 'https://raw.githubusercontent.com/waico/SKAB/master/data/valve1/4.csv',
}

dfs = []
for name, url in data_urls.items():
    file_path = f'../data/{name}.csv'
    if not os.path.exists(file_path):
        urllib.request.urlretrieve(url, file_path)
    df = pd.read_csv(file_path, sep=';')
    dfs.append(df)

# データの結合
df = pd.concat(dfs, ignore_index=True)
print(f"\n結合後のデータ: {len(df)} rows")
display(df.head(3))
display(df.describe().T)

df.to_csv('../data/valve1.csv', index=False)
```
