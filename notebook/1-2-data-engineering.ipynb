## データエンジニアリング

```python
import pandas as pd

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('../data/valve1.csv')
```

```python
df = df.drop(["datetime", "changepoint"], axis=1)
df.head(3)
```

```python
df_x = df.drop('anomaly', axis=1)
#目的変数
df_y = df.anomaly
```

```python
(df_x_train, df_x_test, df_y_train, df_y_test) = train_test_split(df_x, df_y, test_size=0.3, random_state=0)
```

```python
for name, item in {
    'df_x_train': df_x_train,
    'df_x_test': df_x_test,
    'df_y_train': df_y_train,
    'df_y_test': df_y_test
}.items():
    item.to_csv(f'../data/treated-data/{name}.csv', index=False)
```
