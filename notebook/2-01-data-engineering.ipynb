## 2-1. データ収集

本課題では、競馬のレースデータを用いて順位(1位, 3位以内, それ以外)を目的変数とし、馬の体重等のレースデータを説明変数とする問題を実施した。

```python
import pandas as pd

from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../data/unprocessed-data/race_data.csv')
```

```python
df.head()
```

```python
df = df.drop(columns=['Unnamed: 0'])
object_columns = df.select_dtypes(include='object').columns
df = df.drop(columns=object_columns)
df.columns
```

```python
df.info()
```

```python
df.head()
```

```python
treated_df = df.drop(columns=["chakujun","horse_id", "jockey_id", "trainer_id"])
```

```python
treated_df.head()
```

```python
treated_df.describe().T
```

```python
df_x = treated_df.drop('rank_class', axis=1)
#目的変数
df_y = treated_df.rank_class
```

```python
df_x_train, df_x_test, df_y_train, df_y_test = train_test_split(df_x, df_y, test_size=0.3, random_state=42, stratify=df_y)
```

```python
for name, item in {
    'df_x_train': df_x_train,
    'df_x_test': df_x_test,
    'df_y_train': df_y_train,
    'df_y_test': df_y_test
}.items():
    item.to_csv(f'../data/treated-data/task2/{name}.csv', index=False)
```
